# Exploración — Panel de criminalidad (NCRB India)

Análisis exploratorio del panel modelable. El objetivo es predecir si los casos registrados de una categoría de crimen en un estado **aumentan** respecto al año anterior.

```python
import pandas as pd
import matplotlib.pyplot as plt

from src.ingest import construir_panel
from src.config import RAIZ, cargar_params

params = cargar_params()
fuente = RAIZ / params['datos']['fuente_externa']
panel = construir_panel(fuente, params['datos']['metrica_fuente'], anio_corte=None)
print(panel.shape)
panel.head()
```

## Balance de clases y cobertura temporal

```python
print('Balance (1=sube):', round(panel['target'].mean(),3))
panel['year'].value_counts().sort_index().plot(kind='bar', figsize=(10,3), title='Filas por año')
plt.tight_layout(); plt.show()
```

## Señal: reversión a la media

¿El cambio del año previo predice el siguiente?

```python
sub = panel.dropna(subset=['lag2']).copy()
sub['subio_antes'] = sub['delta_prev'] > 0
print(pd.crosstab(sub['subio_antes'], sub['target'], normalize='index').round(3))
```

## Aumentos por categoría de crimen

```python
(panel.groupby('crime_category')['target'].mean().sort_values(ascending=False)
   .plot(kind='barh', figsize=(7,6), title='Proporción de años con aumento por categoría'))
plt.tight_layout(); plt.show()
```
